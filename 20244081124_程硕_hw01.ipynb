## 2.欧⼏⾥得算法

```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m
a=gcd(2018,2019)
b=gcd(1200,300)
c=gcd(5040,60)
print(a,b,c)
```

如果⼀个或两个参数是负数，你的函数会做什么？这种⾏为有意义吗？
加绝对值后，函数会返回两个参数的最大公约数，这有⼀定的数学意义。

## 3.近似欧拉数e

```python
import math

def euler_limit(n):
    return (1 + 1 / n) ** n

def euler_infinite_sum(n):
    total = 0.0
    term = 1.0
    for k in range(n):
        if k > 0:
            term /= k
        total += term
    return total

def euler_approx(tolerance):
    total = 0.0
    term = 1.0
    k = 0
    while abs(total - math.exp(1)) >= tolerance:
        if k > 0:
            term /= k
        total += term
        k += 1
    return total

def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))



# --------- 测试 ---------
print("euler_limit(1000)        =", euler_limit(1000))
print("euler_infinite_sum(0)    =", euler_infinite_sum(0))
print("euler_infinite_sum(1)    =", euler_infinite_sum(1))
print("euler_infinite_sum(4)    =", euler_infinite_sum(4))
print("euler_approx(1e-10)      =", euler_approx(1e-10))
print("math.exp(1)              =", math.exp(1))
print("--- print_euler_sum_table(5) ---")
print_euler_sum_table(5)
```

## 4.测试整数属性

```python
def is_power_of_2(n):
    # 不断除以 2，直到不能整除
    while n % 2 == 0:
        n = n // 2
    # 如果最后剩下 1，说明 n 是 2 的幂
    return n == 1

def is_power(b, n):
    # 不断除以 b，直到不能整除
    while n % b == 0:
        n = n // b
    # 如果最后剩下 1，说明 n 是 b 的幂
    return n == 1
def is_power(b, n):
    if b == 1:
        return n == 1
    while n % b == 0:
        n = n // b
    return n == 1
def test_is_power_of_2():
    print("is_power_of_2(8)  =", is_power_of_2(8))    # True
    print("is_power_of_2(12) =", is_power_of_2(12))   # False


def test_is_power():
    print("is_power(3, 9)  =", is_power(3, 9))    # True
    print("is_power(3, 10) =", is_power(3, 10))   # False


test_is_power_of_2()
test_is_power()
```
