## 实验1.正则表达式的基础匹配

```python
import re
# ---------- 任务1 ----------
text1 = "cat on mat; later: cats and dogs"
print("\n[任务1] 文本:", text1)
print("数字 \\d+ :", re.findall(r"\d+", text1))
print("空白 \\s+ :", re.findall(r"\s+", text1))
print("单词 \\bcat\\b :", re.findall(r"\bcat\b", text1))
# 说明：\\d+ 匹配连续数字；\\s+ 匹配连续空白；\\bcat\\b 只匹配独立单词 cat，
# 不会匹配 cats 中的 cat。

# ---------- 任务2 ----------
text2 = "cat dog bird CAT"
print("\n[任务2] 文本:", text2)
print("cat|dog :", re.findall(r"cat|dog", text2))
# 注意：默认区分大小写，CAT 不会匹配 cat|dog。

s = "aaaa"
print('re.findall(r"a|aa|aaa", "aaaa") :', re.findall(r"a|aa|aaa", s))
print('re.findall(r"aaa|aa|a", "aaaa") :', re.findall(r"aaa|aa|a", s))
# 说明：正则交替从左到右尝试，一旦匹配成功就继续从剩余位置匹配；
# "a|aa|aaa" 优先匹配 a，所以得到 ['a','a','a','a']；
# "aaa|aa|a" 优先匹配 aaa，所以得到 ['aaa','a']。

# ---------- 任务3 ----------
text3 = "ct cat caat caaat"
print("\n[任务3] 文本:", text3)
patterns = [r"ca*t", r"ca+t", r"ca?t", r"ca{2}t", r"ca{1,2}t"]
for p in patterns:
    print(f"{p:10s} ->", re.findall(p, text3))
```

## 实验2：分组、回溯引用与替换

```python
import re

# ---------- 任务1：提取邮箱用户名与域名 ----------
text_email = "联系：alice@example.com；备选：bob@test.org"
print("\n[任务1] 文本:", text_email)
email_pattern = r"([A-Za-z0-9._%+-]+)@([A-Za-z0-9.-]+\.[A-Za-z]{2,})"
matches = re.findall(email_pattern, text_email)
print("用户名与域名:", matches)
for user, domain in matches:
    print(f"  用户名={user}, 域名={domain}")

# ---------- 任务2：后向引用找连续重复单词 ----------
text_dup = "cat cat, dog cat, bird bird"
print("\n[任务2] 文本:", text_dup)
dup_pattern = r"\b(\w+)\s+\1\b"
dups = re.findall(dup_pattern, text_dup)
print("连续重复的单词:", dups)
# 说明：\\1 引用第 1 个捕获组，\\b 保证单词边界。

# ---------- 任务3：日期 年-月-日 -> 月-日-年 ----------
text_date = "2024-05-01\n2024/05/02\n2024.05.03"
print("\n[任务3] 转换前:")
print(text_date)

date_pattern = r"(\d{4})([-/.])(\d{2})\2(\d{2})"
# 捕获组：1=年，2=分隔符，3=月，4=日；\\2 回溯引用保证分隔符一致
new_text = re.sub(date_pattern, r"\3-\4-\1", text_date)

print("\n转换后:")
print(new_text)

# ---------- 任务4：电话号码统一格式 ----------
text_phone = "联系我：123-456-7890 或 987.654.3210"
print("\n[任务4] 文本:", text_phone)
phone_pattern = r"(\d{3})[-.](\d{3})[-.](\d{4})"
new_phone_text = re.sub(phone_pattern, r"(\1) \2-\3", text_phone)
print("统一后:", new_phone_text)
```

## 实验3：使用 BeautifulSoup 解析 HTML

```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")

# ---------- 任务1 ----------
print("\n[任务1] 页面标题文本:", soup.title.get_text())
h1 = soup.find("h1")
print("h1 标签名:", h1.name)
print("h1 文本:", h1.get_text())

# ---------- 任务2 ----------
print("\n[任务2] class='course' 的段落:")
course_ps = soup.find_all("p", class_="course")
for p in course_ps:
    print("  ", p.get_text())

print("所有链接:")
links = soup.find_all("a")
for a in links:
    print(f"  文字={a.get_text()}, href={a.get('href')}")

# ---------- 任务3 ----------
print("\n[任务3] get_text() 提取文本:")
print("整个文档可见文本:")
print(repr(soup.get_text()))

print("\n单个段落文本:")
print(repr(course_ps[0].get_text()))

print("\n单个链接文本:")
print(repr(links[0].get_text()))
```
