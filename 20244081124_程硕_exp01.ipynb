## 1.基本运算、数据类型与字符串

```python
# 1. 运算符含义与结果类型
print("8 / 3  =", 8 / 3, type(8 / 3))    # 真除法，返回 float
print("8 // 3 =", 8 // 3, type(8 // 3))  # 整除（向下取整），返回 int
print("8 % 3  =", 8 % 3, type(8 % 3))    # 取模（余数），返回 int
print("2 ** 7 =", 2 ** 7, type(2 ** 7))  # 幂运算，返回 int

# 2. 四种基本类型
i = 10
f = 3.14
s = "hello"
b = True
print(type(i), type(f), type(s), type(b))

# 3. 类型转换
print(int("123"))    # 字符串 -> 整数 123
print(float("3.14")) # 字符串 -> 浮点数 3.14
print(int(3.9))      # 浮点数 -> 整数，截断小数部分，得 3
try:
    int("goat")
except ValueError as e:
    print("转换失败：", e)  # 字符串不是合法数字，无法转成 int

# 4. 字符串处理
text = " Python Data Analysis "
stripped = text.strip()
print(repr(stripped))
print(stripped.upper())
print(stripped.lower())
print(stripped[0:6])        # 切片
words = stripped.split()    # 拆分单词
print(words)
print("-".join(words))      # 连接

# 5. 拼接、重复与不可变性
print("Py" + "thon")   # 拼接
print("ab" * 3)        # 重复
s2 = "hello"
try:
    s2[0] = "H"        # 字符串不可变，会报错
except TypeError as e:
    print("修改失败：", e)
new_s2 = "H" + s2[1:]  # 构造新字符串
print(new_s2)
```

## 2.函数、条件、循环与递归

```python
# 1. score_level
def score_level(score):
    if score < 0 or score > 100:
        raise ValueError("分数必须在 0~100 之间")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

print(score_level(95), score_level(85), score_level(70), score_level(50))
try:
    score_level(120)
except ValueError as e:
    print("错误：", e)

# 2. for 与 while 求和
total_for = 0
for x in range(1, 101):
    total_for += x

total_while = 0
n = 1
while n <= 100:
    total_while += n
    n += 1

print(total_for, total_while, total_for == total_while)

# 3. 阶乘：递归与迭代
def factorial_rec(n):
    if n < 0:
        raise ValueError("输入必须为非负整数")
    if n == 0:
        return 1
    return n * factorial_rec(n - 1)

def factorial_iter(n):
    if n < 0:
        raise ValueError("输入必须为非负整数")
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result

for n in [0, 1, 5, 10]:
    print(n, factorial_rec(n), factorial_iter(n))

# 4. summary
def summary(values):
    return min(values), max(values), sum(values) / len(values)

mn, mx, avg = summary([85, 58, 92, 76, 60, 89])
print(mn, mx, avg)

# 5. sum_all
def sum_all(*args):
    return sum(args)

print(sum_all())
print(sum_all(5))
print(sum_all(1, 2, 3, 4))
```

## 3.列表、字典、元组与基础统计

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

# 1. zip 构建字典并逐项输出
score_dict = dict(zip(names, scores))
for name, score in score_dict.items():
    print(name, score)

# 2. enumerate，序号从 1 开始
for idx, (name, score) in enumerate(score_dict.items(), start=1):
    print(idx, name, score)

# 3. 列表推导式
passed = [name for name, score in score_dict.items() if score >= 60]
squares = [score ** 2 for score in scores]
print(passed)
print(squares)

# 4. 统计
count = len(scores)
avg = sum(scores) / count
mx = max(scores)
mn = min(scores)
pass_rate = len(passed) / count
print(f"人数={count}, 平均分={avg:.2f}, 最高分={mx}, 最低分={mn}, 及格率={pass_rate:.0%}")

# 5. alias 与 copy 的区别（用单独演示列表）
demo = [1, 2, 3]
alias = demo
copied = demo.copy()
alias[0] = 999
print("demo =", demo)     # 被 alias 修改影响
print("copied =", copied) # 不受影响

# 6. 元组不可变 & zip 长度不同
record = ("Alice", 85)
try:
    record[0] = "Bob"
except TypeError as e:
    print("元组不可修改：", e)

print(list(zip([1, 2, 3], ["a", "b"])))  # 以较短序列为准
```

## 4.文件读写、异常处理与模块

```python
import os
import pickle

# 1. 工作目录与路径
print("当前目录：", os.getcwd())
data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)
file_path = os.path.join(data_dir, "students.txt")

# 2. 写入并读取
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
with open(file_path, "w", encoding="utf-8") as f:
    f.write(content)

with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()
print(lines[:3])

# 3. parse_student
def parse_student(line):
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError("字段数量不正确")
    name, score_str = parts
    if name == "":
        raise ValueError("姓名为空")
    score = int(score_str)  # 非数字会抛 ValueError
    if score < 0 or score > 100:
        raise ValueError("成绩超出 0~100 范围")
    return {"name": name, "score": score}

# 4. 逐行处理，捕获异常
valid_records = []
invalid_count = 0
with open(file_path, "r", encoding="utf-8") as f:
    next(f)  # 跳过表头
    for lineno, line in enumerate(f, start=2):
        try:
            valid_records.append(parse_student(line))
        except ValueError as e:
            invalid_count += 1
            print(f"第 {lineno} 行无效：{e}")

print("有效记录：", valid_records)
print("无效记录数：", invalid_count)

# 5. 写入报告，演示 w 与 a
valid_count = len(valid_records)
avg = sum(r["score"] for r in valid_records) / valid_count
pass_rate = sum(1 for r in valid_records if r["score"] >= 60) / valid_count
report_path = os.path.join(data_dir, "report.txt")

with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数：{valid_count}\n")
    f.write(f"无效记录数：{invalid_count}\n")
    f.write(f"平均分：{avg:.2f}\n")
    f.write(f"及格率：{pass_rate:.0%}\n")

with open(report_path, "a", encoding="utf-8") as f:
    f.write("--- 追加内容 ---\n")  # a 模式在末尾追加，不覆盖

with open(report_path, "r", encoding="utf-8") as f:
    print(f.read())

# 6. pickle 保存与读取
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)

with open(pkl_path, "rb") as f:
    loaded = pickle.load(f)
print("恢复数据与原数据相等：", loaded == valid_records)

# 7. 读取不存在的文件
try:
    with open(os.path.join(data_dir, "no_such_file.txt"), "r", encoding="utf-8") as f:
        f.read()
except FileNotFoundError as e:
    print("文件不存在：", e)

from student_utils import parse_student

print(parse_student("Alice,85"))   # 有效
try:
    parse_student("Henry,105")     # 无效
except ValueError as e:
    print("无效输入：", e)
```

## 5.类、继承与对象复制


```python
import copy

# 1. Student 类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"

# 2. 为有效记录创建对象
valid_records = [
    {"name": "Alice", "score": 85},
    {"name": "Bob", "score": 58},
    {"name": "Carol", "score": 92},
]
students = [Student(r["name"], r["score"]) for r in valid_records]
for s in students:
    print(s, "及格" if s.is_passed() else "不及格")

# 3. 继承
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name}, score={self.score}, topic={self.research_topic})"

g = GraduateStudent("Eva", 90, "机器学习")
print(g, g.is_passed())

# 4. 直接赋值 / 浅拷贝 / 深拷贝
def make_original():
    return {"name": "Alice", "scores": [85, 90]}

# 直接赋值
orig = make_original()
assigned = orig
assigned["scores"].append(100)
print("直接赋值 -> 原对象：", orig)

# 浅拷贝
orig = make_original()
shallow = copy.copy(orig)
shallow["scores"].append(100)
print("浅拷贝 -> 原对象：", orig)   # 内部列表被共享，原对象受影响

# 深拷贝
orig = make_original()
deep = copy.deepcopy(orig)
deep["scores"].append(100)
print("深拷贝 -> 原对象：", orig)   # 完全独立，原对象不受影响
```
