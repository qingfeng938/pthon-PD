## 1.字符串的乐趣

```python

# 字符串处理函数 + 测试
def is_palindrome(s):
    # 去掉空格并转小写，再与反转后的字符串比较
    cleaned = s.replace(' ', '').lower()
    return cleaned == cleaned[::-1]


def is_abecedarian(s):
    # 去掉空格并转小写
    cleaned = s.replace(' ', '').lower()
    # 检查每个字母是否都 <= 后一个字母
    for i in range(len(cleaned) - 1):
        if cleaned[i] > cleaned[i + 1]:
            return False
    return True


def remove_vowels(s):
    # 元音为 a、e、i、o、u（含大写）
    vowels = 'aeiouAEIOU'
    result = ''
    for char in s:
        if char not in vowels:
            result += char
    return result


# ---------------- 测试 ----------------

# 1. is_palindrome
print("=== is_palindrome ===")
print(is_palindrome('tacocat'))            # True
print(is_palindrome('T A C O cat'))        # True
print(is_palindrome('rats live on no evil star'))  # True
print(is_palindrome('hello'))              # False

# 2. is_abecedarian
print("\n=== is_abecedarian ===")
print(is_abecedarian('adder'))             # True
print(is_abecedarian('beet'))              # True
print(is_abecedarian('abcd efgh xyz'))     # True
print(is_abecedarian('dog'))               # False
print(is_abecedarian('cat'))               # False

# 3. remove_vowels
print("\n=== remove_vowels ===")
print(remove_vowels('cat'))                # ct
print(remove_vowels('audio'))              # d
print(remove_vowels('aeiou'))              # （空字符串）
```

## 2.列表的乐趣

```python
# 列表操作函数 + 测试
def list_reverse(lst):
    # 输入不是列表则引发 TypeError
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表")
    # 返回反转后的新列表（不修改原列表）
    return lst[::-1]


def is_sorted(seq):
    # 只遍历一次：比较相邻元素
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False
    return True


# ---------------- 测试 ----------------

# 1. list_reverse
print("=== list_reverse ===")
print(list_reverse([1, 2, 3]))        # [3, 2, 1]
print(list_reverse([]))               # []
print(list_reverse(['a', 'b']))       # ['b', 'a']

# 非列表输入应引发 TypeError
try:
    list_reverse("abc")
except TypeError as e:
    print("TypeError:", e)            # 输入必须是列表

# 2. is_sorted
print("\n=== is_sorted ===")
print(is_sorted([1, 2, 3]))           # True
print(is_sorted([1, 1, 2]))           # True（允许相等）
print(is_sorted([3, 2, 1]))           # False
print(is_sorted([]))                  # True（空序列）
print(is_sorted("abc"))               # True（字符串也是序列）
print(is_sorted((1, 3, 2)))           # False（元组）
```

## 3.作为向量的元组

```python
# 向量与矩阵操作函数 + 测试

def _is_number(x):
    # 判断是否为整数或浮点数（bool 是 int 子类，这里排除掉更严谨）
    return isinstance(x, (int, float)) and not isinstance(x, bool)


def vec_scalar_mult(t, s):
    # 检查 t 是否为元组
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    # 检查 t 的每个元素是否为数字
    for x in t:
        if not _is_number(x):
            raise TypeError("元组元素必须是整数或浮点数")
    # 检查标量 s 是否为数字
    if not _is_number(s):
        raise TypeError("第二个参数必须是整数或浮点数")
    # 返回每个元素乘以 s 的新元组
    return tuple(x * s for x in t)


def vec_inner_product(t1, t2):
    # 检查两个参数是否都是元组
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("两个参数都必须是元组")
    # 检查长度是否一致
    if len(t1) != len(t2):
        raise ValueError("两个向量长度必须一致")
    # 计算内积，返回浮点数
    total = 0.0
    for a, b in zip(t1, t2):
        total += a * b
    return float(total)


def check_valid_mx(mx):
    # 必须是元组
    if not isinstance(mx, tuple):
        return False
    # 每个元素必须是元组
    for row in mx:
        if not isinstance(row, tuple):
            return False
    # 所有内层元组长度相同
    if len(mx) > 0:
        row_len = len(mx[0])
        for row in mx:
            if len(row) != row_len:
                return False
    # 每个元素必须是数字
    for row in mx:
        for x in row:
            if not _is_number(x):
                return False
    return True


def mx_vec_mult(mx, v):
    # 用 check_valid_mx 校验矩阵
    if not check_valid_mx(mx):
        raise ValueError("矩阵不合法")
    # 校验向量
    if not isinstance(v, tuple):
        raise TypeError("向量必须是元组")
    for x in v:
        if not _is_number(x):
            raise TypeError("向量元素必须是整数或浮点数")
    # 矩阵列数必须等于向量长度
    if len(mx) > 0 and len(mx[0]) != len(v):
        raise ValueError("矩阵列数与向量长度不一致")
    # 逐行做内积
    return tuple(vec_inner_product(row, v) for row in mx)


# ---------------- 测试 ----------------

# 1. vec_scalar_mult
print("=== vec_scalar_mult ===")
print(vec_scalar_mult((1, 2, 3), 2))        # (2, 4, 6)
print(vec_scalar_mult((1.5, 2.0), 2))       # (3.0, 4.0)
try:
    vec_scalar_mult([1, 2], 2)
except TypeError as e:
    print("TypeError:", e)

# 2. vec_inner_product
print("\n=== vec_inner_product ===")
print(vec_inner_product((1, 2, 3), (4, 5, 6)))   # 32.0
print(vec_inner_product((1.0, 2.0), (3.0, 4.0))) # 11.0
try:
    vec_inner_product((1, 2), (1, 2, 3))
except ValueError as e:
    print("ValueError:", e)

# 3. check_valid_mx
print("\n=== check_valid_mx ===")
print(check_valid_mx(((1, 2), (3, 4))))     # True
print(check_valid_mx(((1, 2), (3,))))        # False（长度不一致）
print(check_valid_mx(((1, 2), [3, 4])))      # False（内层不是元组）
print(check_valid_mx(((1, 'a'), (3, 4))))    # False（含非数字）

# 4. mx_vec_mult
print("\n=== mx_vec_mult ===")
mx = ((1, 2), (3, 4))
v = (5, 6)
print(mx_vec_mult(mx, v))                    # (17.0, 39.0)
try:
    mx_vec_mult(mx, (1, 2, 3))
except ValueError as e:
    print("ValueError:", e)
```

## 4.更多向量的乐趣

```python

# Vector 类 + 测试

class Vector:
    def __init__(self, dim, elements=None):
        # 校验维度：必须是整数且非负
        if not isinstance(dim, int) or isinstance(dim, bool):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度必须是非负整数")

        # 未提供元素时，默认生成 dim 个 0
        if elements is None:
            elements = [0] * dim

        # 校验元素类型：必须是列表或元组
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
        # 校验每个元素是数字
        for x in elements:
            if not isinstance(x, (int, float)) or isinstance(x, bool):
                raise TypeError("元素必须是整数或浮点数")
        # 校验维度与元素个数一致
        if len(elements) != dim:
            raise ValueError("维度与元素个数不一致")

        # 维度和元素都是实例属性
        self.dim = dim
        self.elements = tuple(elements)

    # 便于打印
    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

    # 比较前先检查维度一致
    def _check_same_dim(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能与 Vector 对象比较")
        if self.dim != other.dim:
            raise ValueError("维度不同的向量不能比较")

    # 相等
    def __eq__(self, other):
        self._check_same_dim(other)
        return self.elements == other.elements

    # 小于（按元组字典序）
    def __lt__(self, other):
        self._check_same_dim(other)
        return self.elements < other.elements

    # 小于或等于
    def __le__(self, other):
        self._check_same_dim(other)
        return self.elements <= other.elements

    # 大于
    def __gt__(self, other):
        self._check_same_dim(other)
        return self.elements > other.elements

    # 大于或等于
    def __ge__(self, other):
        self._check_same_dim(other)
        return self.elements >= other.elements

    # 内积
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是 Vector 对象")
        if self.dim != other.dim:
            raise ValueError("维度不一致，无法计算内积")
        total = 0.0
        for a, b in zip(self.elements, other.elements):
            total += a * b
        return float(total)

    # 乘法：标量 * 向量、向量 * 标量、向量 * 向量
    def __mul__(self, other):
        # 标量乘法
        if isinstance(other, (int, float)) and not isinstance(other, bool):
            return Vector(self.dim, tuple(x * other for x in self.elements))
        # 向量逐元素相乘
        if isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("维度不一致，无法相乘")
            return Vector(self.dim,
                          tuple(a * b for a, b in zip(self.elements, other.elements)))
        raise TypeError("只能与标量或 Vector 相乘")

    # 支持标量左乘：2 * v
    def __rmul__(self, other):
        return self.__mul__(other)


# ---------------- 测试 ----------------

print("=== 构造 ===")
v1 = Vector(2, (2, 3))
v2 = Vector(2, (4, 5))
print(v1)                       # Vector(2, (2, 3))
print(Vector(3))                # Vector(3, (0, 0, 0))  默认全 0

try:
    Vector(-1, ())
except ValueError as e:
    print("ValueError:", e)
try:
    Vector(2, (1, 2, 3))
except ValueError as e:
    print("ValueError:", e)

print("\n=== 比较 ===")
print(Vector(2, (2, 4)) < Vector(2, (2, 5)))   # True
print(Vector(2, (2, 4)) == Vector(2, (2, 4)))  # True
print(Vector(2, (2, 4)) >= Vector(2, (2, 4)))  # True
try:
    Vector(2, (1, 2)) < Vector(3, (1, 2, 3))
except ValueError as e:
    print("ValueError:", e)

print("\n=== 内积 ===")
print(v1.dot(v2))               # 2*4 + 3*5 = 23.0

print("\n=== 乘法 ===")
print(2 * v1)                   # Vector(2, (4, 6))
print(v1 * 2)                   # Vector(2, (4, 6))
print(v1 * v2)                  # Vector(2, (8, 15))
```
